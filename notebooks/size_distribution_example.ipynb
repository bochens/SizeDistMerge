```python
import numpy as np
import sizedistmerge as sdm

edges = np.array([20., 40., 80., 160.])
number_distribution = np.array([1200., 800., 250.])  # dN/dlog10D, cm^-3
counts = sdm.counts_from_dndlog(number_distribution, edges_nm=edges)
print('Particle number in each bin:', counts)
```

## Change diameter without changing the particle number
This simple example applies a uniform diameter factor. Actual optical and
aerodynamic conversions generally vary with diameter.

```python
converted_edges = edges * 1.15
converted_distribution = sdm.remap_dndlog_by_edges(edges, converted_edges, number_distribution)
np.testing.assert_allclose(
    sdm.counts_from_dndlog(converted_distribution, edges_nm=converted_edges), counts)

diameter_aerodynamic = np.array([500., 1000., 2000.])
diameter_volume = sdm.da_to_dv(diameter_aerodynamic, rho_p=1200.)
print('Volume-equivalent diameter (nm):', diameter_volume)
```
